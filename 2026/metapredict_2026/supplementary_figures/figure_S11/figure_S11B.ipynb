# Figure S11B — CAID3 Disorder-PDB by IDR length: APS

APS (average precision) of every CAID3 disorder predictor, scored separately for disordered regions (IDRs) of different lengths. Grey ticks are all 66 predictors; lines are the top five CAID3 methods on the full set and metapredict V1/V2/V3. The numbers above the axes are metapredict V3's rank in each bin.

**How the bins are made.** An IDR is a maximal run of consecutive residues labelled disordered in the CAID3 Disorder-PDB reference. Each bin keeps every ordered residue and only the disordered residues in IDRs of that length; all other disordered residues are excluded. Each bin is then scored exactly as in Figure S3 (pooled per-residue AUC, APS, Fmax, and MCC at each method's Fmax threshold, chosen within the bin), with metapredict predicted by metapredict 3.1.0 (`data/metapredict_caid/`). Short-IDR bins have few disordered residues against the same ordered background, so APS, Fmax and MCC are low there for every method.

**Output:** `figures/figure_S11B.pdf`

```python
import matplotlib.pyplot as plt

from caid3_panels import load_predictions, load_reference
from figtools import apply_style, require_metapredict_3_1, save_panel
from idr_length import describe_bins, field_panel, print_metapredict_ranks_by_bin, rank_by_idr_length

apply_style()
require_metapredict_3_1()
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
```

```python
reference = load_reference()
describe_bins(reference)
rankings = rank_by_idr_length(reference, load_predictions())
print_metapredict_ranks_by_bin(rankings)
```

```python
fig, ax = plt.subplots(figsize=(2.6, 2.2), dpi=150)
field_panel(ax, rankings, 'aps')
save_panel(fig, 'figure_S11B')
```
