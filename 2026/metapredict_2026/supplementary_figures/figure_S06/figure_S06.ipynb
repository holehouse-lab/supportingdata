# Figure S6 — Prediction time vs batch size: all three panels in one figure

Panels A–C (`figure_S06A–C.ipynb`: metapredict V1, V2 and V3) side by side with panel letters, each drawn at the size of its single-panel notebook. Same data and plotting as the panel notebooks (see the README); the legend is drawn in A only, as there.

**Output:** `figures/figure_S06.pdf`

```python
import matplotlib.pyplot as plt

from figtools import apply_style, label_panels, save_panel
from perf_plots import load_results, plot_batch_size_panel

apply_style()
%matplotlib inline
%config InlineBackend.figure_format = 'retina'

VERSIONS = ['V1', 'V2', 'V3']  # panels A-C
PANEL_LETTERS = 'ABC'
# each panel is drawn at the size of the single-panel notebooks (inches)
PANEL_WIDTH_IN = 3.2
PANEL_HEIGHT_IN = 2.6
```

```python
rows = load_results('batch_size')
```

```python
fig, axes = plt.subplots(1, len(VERSIONS), figsize=(PANEL_WIDTH_IN * len(VERSIONS), PANEL_HEIGHT_IN), dpi=150)
for ax, version in zip(axes, VERSIONS):
    plot_batch_size_panel(ax, rows, version, show_legend=(version == VERSIONS[0]))
plt.tight_layout()
label_panels(fig, axes, PANEL_LETTERS)
save_panel(fig, 'figure_S06')
```
