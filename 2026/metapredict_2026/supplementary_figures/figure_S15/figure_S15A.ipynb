# Figure S15A — hallucinated protein 0738

metapredict V3 disorder profile for the de novo designed folded protein 0738 from Anishchenko *et al.* (2021) *Nature* 600, 547–552, as expressed. Shaded regions are predicted IDRs. The dashed line marks the native initiator methionine (residue 21); residues 1–20 are the N-terminal His-tag, which metapredict correctly calls disordered.

**How it is made.** Predicted live with metapredict V3 (`meta.predict_disorder_domains`).

**Output:** `figures/figure_S15A.pdf`

```python
import matplotlib.pyplot as plt

from figtools import apply_style, require_metapredict_3_1, save_panel
from synthetic_disorder import plot_disorder

require_metapredict_3_1()
apply_style()
%matplotlib inline
%config InlineBackend.figure_format = 'retina'

# expressed construct (Anishchenko et al. 2021)
seq_0738 = 'MGSSHHHHHHSSGLVPRGSHMNIQVSLQWEDPKKGKVFSHTVNIPPGGTAEQIADNILDMARSLQDEGWDKLTVQVTVNPGFPKETAMRVAAALKEAFEDRGLRLTSIETSGNSIHLKFRY'
```

```python
fig, ax = plt.subplots(figsize=(3, 1), dpi=300)
plot_disorder(ax, seq_0738, label='0738', start=21)
ax.set_xlabel('Residue', fontsize=8)
plt.tight_layout()
save_panel(fig, 'figure_S15A')
```
