# Figure S15D — designed variants of the hnRNPA1 low-complexity domain

metapredict V3 disorder profiles for wild-type A1-LCD and the five designed variants from Pesce *et al.* (2024) *Sci. Adv.* 10, eadm9926, all of which are experimentally disordered. Shaded regions are predicted IDRs.

**How it is made.** Sequences from `data/synthetic_proteins/pesce_seqs.fasta` in the shared data directory; predicted live with metapredict V3.

**Output:** `figures/figure_S15D.pdf`

```python
import matplotlib.pyplot as plt
import protfasta

from figtools import apply_style, require_metapredict_3_1, save_panel, shared_data_path
from synthetic_disorder import plot_disorder

require_metapredict_3_1()
apply_style()
%matplotlib inline
%config InlineBackend.figure_format = 'retina'

seqs = protfasta.read_fasta(str(shared_data_path('synthetic_proteins', 'pesce_seqs.fasta')))
# wild type first, then variants V1-V5 (Pesce et al. naming)
order = ['WT'] + sorted(k for k in seqs if k != 'WT')
labels = {k: ('wild-type A1-LCD' if k == 'WT' else f'A1-LCD {k}') for k in order}
```

```python
fig, axes = plt.subplots(nrows=len(order), figsize=(3, 0.9 * len(order)), dpi=300, sharex=True)
for ax, k in zip(axes, order):
    plot_disorder(ax, seqs[k], label=labels[k])
axes[-1].set_xlabel('Residue', fontsize=8)
plt.tight_layout()
save_panel(fig, 'figure_S15D')
```
