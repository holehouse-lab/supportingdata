# Figure S10A — A short IDR that metapredict V3 misses: example 1 of 10 (loop)

One full-length CAID3 Disorder-PDB protein. Left, three tracks sharing a residue axis: the metapredict V3 score (3.1.0) with metapredict's own IDR calls shaded and the 0.5 threshold dashed; the AlphaFold2 pLDDT, from the CAID3 AlphaFold-pLDDT predictions (distributed as 1 − pLDDT/100), with the pLDDT = 70 confidence boundary dashed; and the Disorder-PDB reference labels (disordered, ordered, or unscored), with the short IDRs V3 misses outlined. Right, the protein's AlphaFold2 model (AlphaFold DB, `AF-<UniProt>-F1`) as a cartoon in the standard AlphaFold pLDDT colours, with the short IDRs V3 misses drawn as thick, opaque tubes and the rest of the protein translucent. The legend is drawn in this panel only.

**Definitions.** A short IDR is a run of 10–49 residues labelled disordered (CAID3 has none shorter than 10). It is a *loop* if ordered residues flank both sides, a *terminal tail* if it reaches either terminus, and otherwise *next to unscored* residues (residues that are neither DisProt-disordered nor resolved in the PDB). V3 misses an IDR when fewer than half of its residues score ≥ 0.5 and fewer than half fall inside a metapredict IDR call.

**How the ten examples are chosen.** Proteins of at most 600 residues with a missed short IDR are assigned the context holding most of their missed short-IDR residues, and within each context are ranked by missed short-IDR residues. The top five loop, top three next-to-unscored and top two terminal-tail proteins are shown, matching each context's share of the short-IDR residues V3 misses on CAID3 (53%, 26% and 21%).

**Structure render.** Made by `python render_structures.py` (PyMOL), which downloads the model, checks its sequence is identical to the CAID3 sequence, and records in `data/alphafold_renders/models.tsv` how closely its pLDDT matches CAID3's (CAID3 used an earlier AlphaFold DB release). The model is oriented on its core (pLDDT ≥ 50) with the missed short IDRs turned toward the viewer; low-confidence tails can leave the frame.

**Output:** `figures/figure_S10A.pdf`

```python
from caid3_panels import load_predictions, load_reference
from figtools import apply_style, require_metapredict_3_1, save_panel
from short_idr_examples import (plddt_summary, describe_example, describe_selection, example_panel,
                                select_examples)

apply_style()
require_metapredict_3_1()
%matplotlib inline
%config InlineBackend.figure_format = 'retina'

EXAMPLE_INDEX = 0  # panels A–J show examples 0–9
```

```python
reference = load_reference()
predictions = load_predictions(['Metapredict-v3', 'AlphaFold-pLDDT'])
examples = select_examples(reference, predictions['Metapredict-v3'])
describe_selection(examples)
```

Every short IDR of this protein: the fraction of residues V3 scores ≥ 0.5, the fraction inside a metapredict IDR call, and the mean AlphaFold2 pLDDT (CAID3 values).

```python
example = examples[EXAMPLE_INDEX]
describe_example(example, reference, predictions['AlphaFold-pLDDT'][example.target])
```

Across all 319 CAID3 targets: how often AlphaFold2 confidently folds (mean pLDDT > 70) the short IDRs V3 misses, compared with those it finds.

```python
plddt_summary(reference, predictions['Metapredict-v3'], predictions['AlphaFold-pLDDT'])
```

```python
fig = example_panel(example, reference, predictions['Metapredict-v3'], predictions['AlphaFold-pLDDT'],
                    show_legend=True)
save_panel(fig, 'figure_S10A')
```
