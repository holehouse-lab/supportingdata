# Figure S8 — Streaming prediction bounds memory: both panels in one figure

Panels A (`figure_S08A.ipynb`, peak RAM) and B (`figure_S08B.ipynb`, throughput) side by side with panel letters, each drawn at the size of its single-panel notebook. Same data and plotting as the panel notebooks (see the README).

**Output:** `figures/figure_S08.pdf`

```python
import matplotlib.pyplot as plt

from figtools import apply_style, label_panels, save_panel
from perf_plots import load_results, plot_stream_memory, plot_stream_throughput, stream_sweeps

apply_style()
%matplotlib inline
%config InlineBackend.figure_format = 'retina'

PANEL_LETTERS = 'AB'
# each panel is drawn at the size of the single-panel notebooks (inches)
PANEL_WIDTH_IN = 4.1
PANEL_HEIGHT_IN = 3.1
```

```python
sweeps = stream_sweeps(load_results('stream'))
```

```python
fig, (ax_memory, ax_throughput) = plt.subplots(1, 2, figsize=(PANEL_WIDTH_IN * 2, PANEL_HEIGHT_IN), dpi=150)
plot_stream_memory(ax_memory, sweeps)
plot_stream_throughput(ax_throughput, sweeps)
plt.tight_layout()
label_panels(fig, [ax_memory, ax_throughput], PANEL_LETTERS)
save_panel(fig, 'figure_S08')
```
