# Figure S7B — V2: batch vs one-at-a-time prediction

Time for metapredict V2 to predict random subsets of the human proteome, either one sequence at a time (sequential, red) or in a single batched call (batch; black on a CPU, blue on a GPU). One marker shape per machine. Points are measurements; dashed lines are linear fits. Single sequences always run on the CPU; batched calls use metapredict 3.1.0's default batch size for V2 on the device shown in the legend. A machine with a GPU is run twice, batching on its CPU and on its GPU; these are separate series, and its sequential runs are summarised by their median.

**How it is made.** Every TSV in `data/benchmarks/batch_vs_sequential/`, written by `benchmarks/batch_vs_sequential/run_batch_vs_sequential.py` (seeded random subsets, the same subset timed both ways). Repeat measurements are summarised by their median.

**Output:** `figures/figure_S07B.pdf`

```python
import matplotlib.pyplot as plt

from figtools import apply_style, save_panel
from perf_plots import load_results, plot_batch_vs_sequential_panel

apply_style()
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
```

```python
rows = load_results('batch_vs_sequential')
```

```python
fig, ax = plt.subplots(figsize=(3.0, 2.3), dpi=150)
plot_batch_vs_sequential_panel(ax, rows, 'V2')
plt.tight_layout()
save_panel(fig, 'figure_S07B')
```
