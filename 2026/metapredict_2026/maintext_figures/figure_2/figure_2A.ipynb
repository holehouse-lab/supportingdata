# Figure 2A — CAID3 Disorder-PDB: ROC curves against the full field

ROC curves (false positive rate against true positive rate) for metapredict V1, V2 and V3 and the five best CAID3 disorder predictors (by AUC), Every other CAID3 disorder predictor is drawn unlabelled in light grey to show the range of the field. The colour key is drawn once, in panel B; each highlighted predictor's AUC and rank are printed below.

**How it is made.** Every CAID3 disorder predictor (binding and linker predictors excluded) is scored on all covered, non-excluded residues of the CAID3 Disorder-PDB reference pooled together, as in the CAID3 assessment. Predictions are the CAID3 submission files in the shared data directory, except metapredict, which is predicted with metapredict 3.1.0 (`data/metapredict_caid/`).

**Output:** `figures/figure_2A.pdf`

```python
import matplotlib.pyplot as plt

from caid3_panels import load_predictions, load_reference, print_metapredict_ranks, rank_predictors
from figtools import apply_style, require_metapredict_3_1, save_panel

apply_style()
require_metapredict_3_1()
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
from caid3_panels import field_curve_panel
```

```python
reference = load_reference()
predictions = load_predictions()
ranking = rank_predictors(reference, predictions)
print(f'{len(ranking)} disorder predictors')
print_metapredict_ranks(ranking)
```

```python
fig, ax = plt.subplots(figsize=(3.1, 2.75), dpi=150, layout='constrained')
field_curve_panel(ax, 'roc', reference, predictions, ranking, legend=False)
save_panel(fig, 'figure_2A')
```
